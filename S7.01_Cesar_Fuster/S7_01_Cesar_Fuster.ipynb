**Algoritmo principal**

La función principal se llamará main_seleccio(), sus tareas son:

1. Crear una lista vacía para los candidatos válidos.
2. Mientras haya menos de 10 candidatos registrados:
- Recibir los datos del candidato.
- Validar dichos datos:
  - En caso de no ser válidos, mostrar un mensaje de error.
  - En caso de ser válidos, calcular su puntuación.
- Clasificar al candidato.
- Mostrar su puntuación final.
- Guardar al candidato en la lista.
3. Cuando haya 10 candidatos válidos, dejar de aceptar nuevas inscripciones.
4. Ordenar la lista por prioridad y puntuación.
5. Mostrar el informe final.

Para realizar dichas tareas se crean diferentes funciones que serán llamadas por main_seleccio():
- obtenir_dades(): solicita los datos del candidato.
- validar_dades(): valida los datos.
- calcular_puntuacio(): calcula la puntuación total.
- classificar_candidat(): clasifica al candidato.
- mostrar_individual(): muestra el resultado de cada candidato.
- mostrar_informe(): ordena y muestra el informe final.

**Desgranado del problema**

obtenir_dades()
1. Pedir el nombre y los apellidos.
2. Pedir el teléfono de contacto.
3. Pedir los años de experiencia.
4. Pedir el nivel de catalán.
5. Pedir el nivel de Excel.
6. Pedir el nivel de titulación académica.
7. Devolver todos los datos recibidos.

Se guardan todos los datos en formato text, en un diccionario llamado dades, para que se puedan validar.

```python
def obtenir_dades():
    nombre = input("Introduce tu nombre y apellidos: ")
    telefono = input("Introduce tu número de teléfono: ")
    experiencia = input("Introduce tus años de experiencia: ")
    catalan = input("Introduce tu nivel de catalán (0-3): ")
    excel = input("Introduce tu nivel de Excel (0-3): ")
    titulacion = input("Introduce tu nivel de titulación (0-3): ")

    dades = {
        "nombre": nombre,
        "telefono": telefono,
        "experiencia": experiencia,
        "catalan": catalan,
        "excel": excel,
        "titulacion": titulacion
    }

    return dades
```

validar_dades()
1. Comprobar que el nombre no esté vacío.
2. Comprobar que el número de teléfono solamente se compone de dígitos.
3. Convertir los años de experiencia y las puntuaciones a números enteros.
4. Comprobar que los años de experiencia sean igual o mayor que 0.
5. Comprobar que las puntuaciones estén entre 0 y 3.
6. Si todos los datos son correctos, devolver los datos validados.
7. Si algún dato es incorrecto, mostrar un mensaje de error y devolver None.

```python
def validar_dades(dades):
    try:
        nombre = dades["nombre"].strip()
        telefono = dades["telefono"].strip()
        #.strip() elimina los espacios al inicio y al final, para asegurar que no se acepte un texto con solo espacios como nombre

        experiencia = int(dades["experiencia"])
        catalan = int(dades["catalan"])
        excel = int(dades["excel"])
        titulacion = int(dades["titulacion"])

        if nombre == "":
            print("Error: El nombre no puede estar vacío.")
            print("   ")
            return None

        if not telefono.isdigit():
            print("Error: El número de teléfono introducido no es válido.")
            print("   ")
            return None

        if experiencia < 0:
            print("Error: La experiencia no puede ser negativa.")
            print("   ")
            return None

        if not (0 <= catalan <= 3):
            print("Error: El nivel de catalán debe estar entre 0 y 3.")
            print("   ")
            return None

        if not (0 <= excel <= 3):
            print("Error: El nivel de Excel debe estar entre 0 y 3.")
            print("   ")
            return None

        if not (0 <= titulacion <= 3):
            print("Error: El nivel de titulación debe estar entre 0 y 3.")
            print("   ")
            return None

        dades["nombre"] = nombre
        dades["telefono"] = telefono
        dades["experiencia"] = experiencia
        dades["catalan"] = catalan
        dades["excel"] = excel
        dades["titulacion"] = titulacion

        return dades

    except ValueError:
        print("Error: Los valores numéricos introducidos deben ser números enteros.")
        print("   ")
        return None
```

calcular_puntuacio()
1. Multiplicar los años de experiencia por 1,5.
2. Sumar los puntos de catalán, Excel y titulación.
3. Devolver la puntuación total.

```python
def calcular_puntuacio(dades):
    puntuacion = (dades["experiencia"] * 1.5 + dades["catalan"] + dades["excel"] + dades["titulacion"])
    return puntuacion
```

classificar_candidat()

Tomando la puntuación de la anterior función:
1. Si es mayor o igual que 12, asignar "Prioridad alta".
2. Si es menor de 12 y mayor o igual que 8, asignar «Candidato adecuado».
3. Si no se cumplen las anteriores, asignar «No prioritario».
4. Devolver la clasificación.

```python
def classificar_candidat(puntuacion):
    if puntuacion >= 12:
        classificacion = "Prioridad alta"
    elif 12 > puntuacion >= 8:
        classificacion = "Candidato adecuado"
    else:
        classificacion = "No prioritario"

    return classificacion
```

mostrar_individual()
1. Mostrar nombre y apellidos del candidato.
2. Mostrar su puntuación.
3. Mostrar la clasificación recibida.

```python
def mostrar_individual(dades):
  print("   ")
  print("RESULTADO INDIVIDUAL")
  print(f"Nombre: {dades['nombre']}")
  print(f"Puntuación: {dades['puntuacion']}")
  print(f"Clasificación: {dades['classificacion']}")
  print("   ")  #Espacio entre candidatos
```

mostrar_informe()
1. Ordenar los candidatos de mayor a menor puntuación.
2. Mostrar nombre y apellidos, teléfono, puntuación y clasificación de cada candidato.

```python
def mostrar_informe(candidatos):
    candidatos.sort(
        key=lambda candidato: candidato["puntuacion"],
        reverse=True
    ) #key necesita una función pero lambda permite usarla sin tener que crear una

    print("   ")
    print("INFORME DE CANDIDATOS")

    for candidato in candidatos:
        print(f"Nombre: {candidato['nombre']}")
        print(f"Teléfono: {candidato['telefono']}")
        print(f"Puntuación: {candidato['puntuacion']}")
        print(f"Clasificación: {candidato['classificacion']}")
        print("   ")  #Espacio entre candidatos
```

**main_seleccio()**

Coordina las funciones creadas para generar la lista de candidatos y mostrar el informe resultante.

```python
def main_seleccio():
    candidatos = []

    while len(candidatos) < 10:
        dades = obtenir_dades()
        dades = validar_dades(dades)

        if dades is not None:
            puntuacion = calcular_puntuacio(dades)
            classificacion = classificar_candidat(puntuacion)

            dades["puntuacion"] = puntuacion
            dades["classificacion"] = classificacion

            mostrar_individual(dades)

            candidatos.append(dades)

    mostrar_informe(candidatos)
```

3. Crear un diccionario con los factores correspondientes a cada código de comportamiento, e incluir posteriormente el segmento y mensaje.
4. Identificar el segmento de cada cliente según el código de comportamiento y asignar un mensaje. Se crea un mensaje para cada segmento, ya que no se ha proporcionado.

6. Guardar los resultados en una estructura de datos que contenga la información de cada cliente, además de su segmento, mensaje, factores y riesgo final.

**Algoritmo principal**

La función principal se llamará main_marketing(), y utilizará las siguientes funciones:
1. processar_clients(): toma los datos de los clientes y gestiona la segmentación, los factores, y el cálculo de riesgo final. También asigna el mensaje a cada segmento.
- obtenir_segment_i_factor(): establece el segmento y su mensaje, e indica el factor de comportamiento.
- obtenir_factor_edat(): indica el factor del grupo de edad.
- calcular_risc(): realiza el cálculo del riesgo final.
2. mostrar_dataframe(): muestra los resultados de los clientes procesados.
3. mostrar_respostes(): analiza los resultados y muestra las respuestas a las preguntas de marketing.

**Desgranado del problema**

obtenir_segment_i_factor()
- Se toma el código de comportamiento del cliente, y se devuelve un diccionario con el segmento, el factor y el mensaje de dicho código.

```python
def obtenir_segment_i_factor(codigo_comportamiento):
    dicc_comportamiento = {
        0: {
            "segmento": "Navega pero no añade productos",
            "factor": 0.10,
            "mensaje": "Encuentra lo que buscas entre todos nuestros productos!"
        },
        1: {
            "segmento": "Añade productos, pero no llega al checkout",
            "factor": 0.25,
            "mensaje": "Tus productos favoritos te esperan. Echa un vistazo a tu cesta!"
        },
        2: {
            "segmento": "Llega al checkout, pero no paga",
            "factor": 0.50,
            "mensaje": "Ya casi lo tienes! Completa tu compra de forma sencilla"
        },
        3: {
            "segmento": "Cesta abandonada hace menos de 24 horas",
            "factor": 0.75,
            "mensaje": "Has dejado productos en tu cesta. Vuelve para terminar tu compra!"
        },
        4: {
            "segmento": "Cesta abandonada hace más de 24 horas",
            "factor": 0.90,
            "mensaje": "No te olvides de tu cesta! Tus productos siguen esperándote"
        }
    }

    return dicc_comportamiento[codigo_comportamiento]
```

obtenir_factor_edat()
- Se toma el grupo de edad y se devuelve el factor correspondiente.

```python
def obtenir_factor_edat(grupo_edad):
    factores_edad = {
        "18-25": 1.20,
        "26-40": 1.00,
        "41-60": 0.85,
        "60+": 0.70
    }

    return factores_edad[grupo_edad]
```

calcular_risc()
- Se toman ambos factores y se multiplican para recibir el riesgo final.

```python
def calcular_risc(factor_comportamiento, factor_edad):
    riesgo_final = factor_comportamiento * factor_edad

    return riesgo_final
```

processar_clients()
- Se aplican las funciones ya creadas para cada cliente del diccionario de clientes proporcionado.
- Se guardan tanto los datos de los clientes como los resultados de las funciones en un nuevo diccionario.

```python
def processar_clients(clientes):
    clientes_procesados = {}

    for id_cliente, cliente in clientes.items():

        datos_comportamiento = obtenir_segment_i_factor(cliente["codigo_comportamiento"])

        factor_edad = obtenir_factor_edat(cliente["grupo_edad"])

        riesgo_final = calcular_risc(datos_comportamiento["factor"], factor_edad)

        clientes_procesados[id_cliente] = {
            "id_cliente": id_cliente,
            "nombre": cliente["nombre"],
            "edad": cliente["edad"],
            "grupo_edad": cliente["grupo_edad"],
            "codigo_comportamiento": cliente["codigo_comportamiento"],
            "segmento": datos_comportamiento["segmento"],
            "mensaje": datos_comportamiento["mensaje"],
            "factor_comportamiento": datos_comportamiento["factor"],
            "factor_edad": factor_edad,
            "riesgo_final": riesgo_final
        }

    return clientes_procesados
```

mostrar_dataframe()
- Muestra uno por uno todos los datos proporcionados de cada cliente.
- Muestra también su segmento, mensaje, factor de comportamiento, factor de edad, y riesgo final.

```python
def mostrar_dataframe(clientes_procesados):
    for cliente in clientes_procesados.values():
        print("\nid:", cliente["id_cliente"])
        print("Nombre:", cliente["nombre"])
        print("Edad:", cliente["edad"])
        print("Grupo de edad:", cliente["grupo_edad"])
        print("Código de comportamiento:", cliente["codigo_comportamiento"])
        print("Segmento:", cliente["segmento"])
        print("Mensaje:", cliente["mensaje"])
        print("Factor de comportamiento:", cliente["factor_comportamiento"])
        print("Factor de edad:", cliente["factor_edad"])
        print("Riesgo final:", round(cliente["riesgo_final"], 2))
```

mostrar_respostes()

Se responden las preguntas de marketing en orden:
1. Clientes a contactar primero por la campaña de marketing:
- Se muestran los clientes ordenados por riesgo final de mayor a menor.

2. Qué diferencias hay entre clientes con el mismo comportamiento pero de edades diferentes:

- Se muestran los clientes con código de comportamiento 0.

4. Qué cliente sería el más rentable de recuperar:
- Se muestra al cliente con un mayor riesgo final.

```python
def mostrar_respostes(clientes_procesados):

# 1. Clientes a contactar primero por la campaña de marketing:
    clientes_ordenados = sorted(clientes_procesados.values(), key=lambda cliente: cliente["riesgo_final"], reverse=True)

    print("\n1. Clientes a contactar primero por la campaña de marketing?")

    for cliente in clientes_ordenados:
        print(cliente["nombre"], "- Riesgo:", round(cliente["riesgo_final"], 2))


# 2. Qué diferencias hay entre clientes con el mismo comportamiento pero de edades diferentes:
    print("\n2. Qué diferencias hay entre clientes con el mismo comportamiento pero de edades diferentes:")

    for codigo in range(5):
      print("\nClientes con código de comportamiento", codigo)

      for cliente in clientes_ordenados:
          if cliente["codigo_comportamiento"] == codigo:
              print("Nombre:", cliente["nombre"], "- Grupo de edad:", cliente["grupo_edad"],"- Riesgo final:", round(cliente["riesgo_final"], 2))


# 3. Clientes que muestran tan poco interés que vale la pena invertir presupuesto:
    print("\n3. Clientes que muestran tan poco interés que vale la pena invertir presupuesto:")

    clientes_menor_interes = [cliente for cliente in clientes_ordenados if cliente["codigo_comportamiento"] == 0]

    for cliente in clientes_menor_interes:
        print(cliente["nombre"], "- Riesgo:", round(cliente["riesgo_final"], 2))


# 4. Qué cliente sería el más rentable de recuperar:
    print("\n4. Qué cliente sería el más rentable de recuperar:")

    cliente_prioritario = clientes_ordenados[0]

    print("Nombre:", cliente_prioritario["nombre"])
    print("Segmento:", cliente_prioritario["segmento"])
    print("Riesgo:", round(cliente_prioritario["riesgo_final"], 2))
```

**Algoritmo Inicial**

**Desgranado del problema**

estructurar_paciente()
- Recibe los datos brutos y devuelve un diccionario con nombre, edad y síntomas (normalizados).

```python
def estructurar_paciente(registro):
    datos = registro.split("|")

    nombre = datos[0].strip()
    edad = int(datos[1].strip())
    sintomas = normalizar_sintomas(datos[2])

    return {"nombre": nombre, "edad": edad, "sintomas": sintomas}
```

normalizar_sintomas()
- Recibe la cadena de texto con los síntomas y la procesa para devolver una lista de síntomas normalizados.

```python
def normalizar_sintomas(texto_sintomas):
    sintomas = texto_sintomas.split(",")
    sintomas_normalizados = []

    for sintoma in sintomas:
        sintoma = sintoma.strip().lower()
        sintoma = sintoma.replace(" ", "_")

        if sintoma:
            sintomas_normalizados.append(sintoma)
            #Guarda el síntoma si no está vacío

    return sintomas_normalizados
```

calcular_riesgo
- Toma la lista de síntomas del paciente y el diccionario de niveles de riesgo para calcular el nivel de riesgo total del paciente.

```python
def calcular_riesgo(sintomas, niveles_riesgo):
    riesgo_total = 0

    for sintoma in sintomas:
        riesgo_total += niveles_riesgo[sintoma]

    return riesgo_total
```

procesar_pacientes()
- Crea un diccionario de pacientes procesados con los datos estructurados de cada paciente y su riesgo total.

```python
def procesar_pacientes(pacientes, niveles_riesgo):
    pacientes_procesados = {}

    for registro in pacientes:
        paciente = estructurar_paciente(registro)

        nombre = paciente["nombre"]
        edad = paciente["edad"]
        sintomas = paciente["sintomas"]

        riesgo_total = calcular_riesgo(sintomas, niveles_riesgo)

        paciente["riesgo_total"] = riesgo_total
        pacientes_procesados[nombre] = paciente

    return pacientes_procesados
```

ordenar_pacientes()
- Toma el diccionario de pacientes procesados y los ordena según su riesgo total de mayor a menor.


```python
def ordenar_pacientes(pacientes_procesados):
    pacientes_ordenados = sorted(pacientes_procesados.values(), key=lambda paciente: paciente["riesgo_total"], reverse=True)

    return pacientes_ordenados
```

mostrar_resultados()
- Toma los datos de los pacientes ya ordenados y los imprime junto con un contador de posicion para definir el orden de prioridad.

```python
def mostrar_resultados(pacientes_ordenados):
    print("\nPacientes en orden prioritario de atención:")

    posicion = 0

    for paciente in pacientes_ordenados:
        posicion += 1
        print("-----------------------------------")
        print("Prioridad:", posicion)
        print("Nombre:", paciente["nombre"])
        print("Edad:", paciente["edad"])
        print("Síntomas:", paciente["sintomas"])
        print("Riesgo total:", paciente["riesgo_total"])
```

precio_final()

```python
def precio_final(precio_base, edad):
    resultado = precio_base * (1 + edad / 100)

    return resultado
```
